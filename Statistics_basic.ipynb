```python
## 2.What are the measures of central tendency, and when should you use each? Discuss the mean, median,and mode with examples and situations where each is appropriate.
```

```python
## Mean: Sum of all values divided by number of values.
## Median: Middle value when sorted.
## Mode: Most frequent value.

## When to use:
## Mean: Symmetrical data.
## Median: Skewed data.
## Mode: Categorical data.

import numpy as np
from scipy import stats

data = [10, 20, 20, 30, 40, 50]
print(f"Mean: {np.mean(data)}")
print(f"Median: {np.median(data)}")
print(f"Mode: {stats.mode(data, keepdims=True).mode[0]}")
```

```python
## 3.Explain the concept of dispersion. How do variance and standard deviation measure the spread of data?
```

```python
"""
- Dispersion: How spread out the data values are.
- Variance: Average squared difference from the mean.
- Standard Deviation: Square root of variance.
"""

print(f"Variance: {np.var(data)}")
print(f"Standard Deviation: {np.std(data)}")
```

```python
## 4.What is a box plot, and what can it tell you about the distribution of data?
```

```python
"""
- Box Plot: Shows distribution of data based on min, Q1, median, Q3, max.
- The box shows Q1 to Q3 with a line at the median. Whiskers extend to minimum and maximum values.
"""

import matplotlib.pyplot as plt

plt.boxplot(data)
plt.title('Box Plot')
plt.show()
```

```python
## 6.Explain the concept of skewness and its types.How does skewness aftect the interpretation of data?
```

```python
"""
- Skewness: Measures asymmetry of data.
  - Positive Skew: Right-tailed.
  - Negative Skew: Left-tailed.
  - Zero Skew: Symmetrical.
"""

from scipy.stats import skew

print(f"Skewness: {skew(data)}")
```

```python
## 7.What is the interquartile range (IQR), and how is it used to detect outliers?
```

```python
"""
- Interquartile Range (IQR) = Q3 - Q1.
- Detect outliers:
    Lower bound = Q1 - 1.5 * IQR
    Upper bound = Q3 + 1.5 * IQR
"""

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"IQR: {IQR}")
print(f"Outlier Range: {lower_bound} to {upper_bound}")
```
